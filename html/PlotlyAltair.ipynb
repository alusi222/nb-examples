# Task
Load a dataset from sklearn, then create 10 charts each using matplotlib, altair, and plotly to visualize the data.

## Load data

### Subtask:
Load a sample dataset from the scikit-learn library.

**Reasoning**:
Import the `load_iris` function from `sklearn.datasets` and load the iris dataset.


```python
from sklearn.datasets import load_iris
iris = load_iris()
```

## Explore data

### Subtask:
Display the first few rows of the dataset and its description to understand its structure and content.

**Reasoning**:
Display the first few rows of the dataset and its description to understand its structure and content.


```python
import pandas as pd
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print("First 5 rows of the dataset:")
display(df.head())
print("\nDescription of the dataset:")
print(iris.DESCR)
```

## Visualize with matplotlib

### Subtask:
Create 10 different plots using Matplotlib to visualize the data.

**Reasoning**:
Generate and display 10 different plots using Matplotlib to visualize the data in the DataFrame `df`, including appropriate labels and titles for each plot.


```python
import matplotlib.pyplot as plt

# Plot 1: Scatter plot of sepal length vs. sepal width
plt.figure(figsize=(8, 6))
plt.scatter(df['sepal length (cm)'], df['sepal width (cm)'])
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length vs. Sepal Width')
plt.show()

# Plot 2: Scatter plot of petal length vs. petal width
plt.figure(figsize=(8, 6))
plt.scatter(df['petal length (cm)'], df['petal width (cm)'])
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Length vs. Petal Width')
plt.show()

# Plot 3: Histogram of sepal length
plt.figure(figsize=(8, 6))
plt.hist(df['sepal length (cm)'], bins=10)
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Sepal Length')
plt.show()

# Plot 4: Histogram of sepal width
plt.figure(figsize=(8, 6))
plt.hist(df['sepal width (cm)'], bins=10)
plt.xlabel('Sepal Width (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Sepal Width')
plt.show()

# Plot 5: Histogram of petal length
plt.figure(figsize=(8, 6))
plt.hist(df['petal length (cm)'], bins=10)
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Petal Length')
plt.show()

# Plot 6: Histogram of petal width
plt.figure(figsize=(8, 6))
plt.hist(df['petal width (cm)'], bins=10)
plt.xlabel('Petal Width (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Petal Width')
plt.show()

# Plot 7: Box plot of sepal length
plt.figure(figsize=(8, 6))
plt.boxplot(df['sepal length (cm)'])
plt.ylabel('Sepal Length (cm)')
plt.title('Box Plot of Sepal Length')
plt.show()

# Plot 8: Box plot of sepal width
plt.figure(figsize=(8, 6))
plt.boxplot(df['sepal width (cm)'])
plt.ylabel('Sepal Width (cm)')
plt.title('Box Plot of Sepal Width')
plt.show()

# Plot 9: Box plot of petal length
plt.figure(figsize=(8, 6))
plt.boxplot(df['petal length (cm)'])
plt.ylabel('Petal Length (cm)')
plt.title('Box Plot of Petal Length')
plt.show()

# Plot 10: Box plot of petal width
plt.figure(figsize=(8, 6))
plt.boxplot(df['petal width (cm)'])
plt.ylabel('Petal Width (cm)')
plt.title('Box Plot of Petal Width')
plt.show()
```

## Visualize with altair

### Subtask:
Create 10 different plots using Altair to visualize the data.

**Reasoning**:
Import the altair library and create the requested Altair charts, displaying each one.


```python
import altair as alt

# Plot 1: Scatter plot of 'sepal length (cm)' vs. 'sepal width (cm)'
chart1 = alt.Chart(df).mark_point().encode(
    x='sepal length (cm)',
    y='sepal width (cm)'
).properties(
    title='Sepal Length vs. Sepal Width (Altair)'
)
display(chart1)

# Plot 2: Scatter plot of 'petal length (cm)' vs. 'petal width (cm)'
chart2 = alt.Chart(df).mark_point().encode(
    x='petal length (cm)',
    y='petal width (cm)'
).properties(
    title='Petal Length vs. Petal Width (Altair)'
)
display(chart2)

# Plot 3: Histogram of 'sepal length (cm)'
chart3 = alt.Chart(df).mark_bar().encode(
    x=alt.X('sepal length (cm)', bin=True),
    y='count()'
).properties(
    title='Distribution of Sepal Length (Altair)'
)
display(chart3)

# Plot 4: Histogram of 'sepal width (cm)'
chart4 = alt.Chart(df).mark_bar().encode(
    x=alt.X('sepal width (cm)', bin=True),
    y='count()'
).properties(
    title='Distribution of Sepal Width (Altair)'
)
display(chart4)

# Plot 5: Histogram of 'petal length (cm)'
chart5 = alt.Chart(df).mark_bar().encode(
    x=alt.X('petal length (cm)', bin=True),
    y='count()'
).properties(
    title='Distribution of Petal Length (Altair)'
)
display(chart5)

# Plot 6: Histogram of 'petal width (cm)'
chart6 = alt.Chart(df).mark_bar().encode(
    x=alt.X('petal width (cm)', bin=True),
    y='count()'
).properties(
    title='Distribution of Petal Width (Altair)'
)
display(chart6)

# Plot 7: Box plot of 'sepal length (cm)'
chart7 = alt.Chart(df).mark_boxplot().encode(
    y='sepal length (cm)'
).properties(
    title='Box Plot of Sepal Length (Altair)'
)
display(chart7)

# Plot 8: Box plot of 'sepal width (cm)'
chart8 = alt.Chart(df).mark_boxplot().encode(
    y='sepal width (cm)'
).properties(
    title='Box Plot of Sepal Width (Altair)'
)
display(chart8)

# Plot 9: Box plot of 'petal length (cm)'
chart9 = alt.Chart(df).mark_boxplot().encode(
    y='petal length (cm)'
).properties(
    title='Box Plot of Petal Length (Altair)'
)
display(chart9)

# Plot 10: Box plot of 'petal width (cm)'
chart10 = alt.Chart(df).mark_boxplot().encode(
    y='petal width (cm)'
).properties(
    title='Box Plot of Petal Width (Altair)'
)
display(chart10)
```

## Visualize with plotly

### Subtask:
Create 10 different plots using Plotly to visualize the data.

**Reasoning**:
Import plotly.express and create the first two scatter plots as per the instructions.


```python
import plotly.express as px

# Plot 1: Scatter plot of sepal length vs. sepal width
fig1 = px.scatter(df, x='sepal length (cm)', y='sepal width (cm)',
                  title='Sepal Length vs. Sepal Width (Plotly)')
fig1.show()

# Plot 2: Scatter plot of petal length vs. petal width
fig2 = px.scatter(df, x='petal length (cm)', y='petal width (cm)',
                  title='Petal Length vs. Petal Width (Plotly)')
fig2.show()
```

**Reasoning**:
Create and display the four histograms as per the instructions.


```python
# Plot 3: Histogram of sepal length
fig3 = px.histogram(df, x='sepal length (cm)',
                    title='Distribution of Sepal Length (Plotly)')
fig3.show()

# Plot 4: Histogram of sepal width
fig4 = px.histogram(df, x='sepal width (cm)',
                    title='Distribution of Sepal Width (Plotly)')
fig4.show()

# Plot 5: Histogram of petal length
fig5 = px.histogram(df, x='petal length (cm)',
                    title='Distribution of Petal Length (Plotly)')
fig5.show()

# Plot 6: Histogram of petal width
fig6 = px.histogram(df, x='petal width (cm)',
                    title='Distribution of Petal Width (Plotly)')
fig6.show()
```

**Reasoning**:
Create and display the four box plots as per the instructions.


```python
# Plot 7: Box plot of sepal length
fig7 = px.box(df, y='sepal length (cm)',
              title='Box Plot of Sepal Length (Plotly)')
fig7.show()

# Plot 8: Box plot of sepal width
fig8 = px.box(df, y='sepal width (cm)',
              title='Box Plot of Sepal Width (Plotly)')
fig8.show()

# Plot 9: Box plot of petal length
fig9 = px.box(df, y='petal length (cm)',
              title='Box Plot of Petal Length (Plotly)')
fig9.show()

# Plot 10: Box plot of petal width
fig10 = px.box(df, y='petal width (cm)',
               title='Box Plot of Petal Width (Plotly)')
fig10.show()
```
